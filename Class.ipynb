```python
import random
from datetime import datetime, timedelta


class Book:

    def __init__(self, book_id, title, author, category):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.category = category
        self.is_available = True  # สถานะ: โดนยืม / อยู่ในคลัง

    def borrow(self):
        """เปลี่ยนสถานะเป็นถูกยืม"""
        self.is_available = False

    def return_book(self):
        """เปลี่ยนสถานะกลับคืนคลัง"""
        self.is_available = True


class Member:

    def __init__(self, member_id, name, member_type="นักศึกษา"):
        self.member_id = member_id
        self.name = name
        self.member_type = member_type  # "นักศึกษา" หรือ "อาจารย์"

    def get_loan_period(self):
        """คืนจำนวนวันที่ยืมได้ ตามประเภทสมาชิก"""
        return 30 if self.member_type == "อาจารย์" else 14


class BookLoan:
    FINE_PER_DAY = 5  # บาท/วัน

    def __init__(self, loan_id, book: Book, member: Member, borrow_date):
        self.loan_id = loan_id
        self.book = book
        self.member = member
        self.borrow_date = borrow_date
        self.due_date = borrow_date + timedelta(days=member.get_loan_period())
        self.return_date = None  # ยังไม่คืนตอนสร้าง

    def calculate_late_fee(self):
        """คำนวณค่าปรับ หากเกินกำหนด"""
        if self.return_date and self.return_date > self.due_date:
            late_days = (self.return_date - self.due_date).days
            return late_days * self.FINE_PER_DAY
        return 0

    def mark_returned(self, return_date):
        """บันทึกการคืนว่าเมื่อไหร่"""
        self.return_date = return_date
        self.book.return_book()
```
